```python
import pandas as pd
```

```python
raw = pd.read_csv(r'C:\Users\Denzen Computer\Downloads\teaching_prices_messy.csv')
```

```python
clean = raw.copy()
```

```python
print("=== BEFORE CLEANING ===")
print(f"Rows: {len(clean)}")
print(f"Duplicates: {clean.duplicated().sum()}")
print(f"Missing prices: {clean['price'].isna().sum()}")
print(f"Price column type: {clean['price'].dtype}")
print("\nItem names (notice the lowercase 'oil b'):")
print(clean['item'].value_counts())
```

```python
clean['item'] = clean['item'].str.strip().str.title()
```

```python
clean['price'] = clean['price'].astype(str).str.replace(',', '', regex=False)
clean['price'] = pd.to_numeric(clean['price'], errors='coerce')
```

```python
clean.loc[clean['price'] == 9999, 'price'] = float('nan')
```

```python
clean = clean.drop_duplicates().reset_index(drop=True)
```

```python
print("\n=== AFTER CLEANING ===")
print(f"New Shape: {clean.shape}")
print(f"Price column type: {clean['price'].dtype}")
print("\nMissing values now (should be 2: the original missing + the flagged 9999):")
print(clean.isna().sum())
print("\nItem names (should all be Title Case now):")
print(clean['item'].value_counts())
```
